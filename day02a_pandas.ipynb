## The one idea to hold on to

A **DataFrame** is a table: rows and named columns, like a spreadsheet or a SQL table.

A **Series** is a single column of that table.

`df["age"]` gives you a Series. `df[["age", "fare"]]` — note the double brackets — gives you a DataFrame with two columns. Nearly every confusing Pandas error traces back to this distinction.

````python
import seaborn as sns
import pandas as pd

df = sns.load_dataset("titanic")

print(type(df[["age", "fare"]]))   # DataFrame — double brackets
print(type(df["age"]))             # Series — single brackets
````

## 3. Find the missing values

This single line is the first thing to run on any new dataset.

````python
df.isnull().sum().sort_values(ascending=False)
````

Read it inside out: `isnull()` turns every cell into True/False, `sum()` counts the Trues per column (True counts as 1), `sort_values()` puts the worst offenders on top.

---
## 4. Deal with the missing values

````python
print("Before:", df.shape)

# Option A: throw away rows that have any missing value
print("After dropna():", df.dropna().shape)

# Only look at the columns you care about — far less destructive
cols = ["age", "fare", "pclass", "survived"]
print("After dropna() on 4 columns:", df[cols].dropna().shape)
````

````python
# Option B: fill the gaps instead of dropping rows
filled = df["age"].fillna(df["age"].median())
print("Missing before:", df["age"].isnull().sum(), "| after:", filled.isnull().sum())
print("Filled with:", df["age"].median())
````

## 5. Filter rows by a condition

````python
children = df[df["age"] < 18]
print("Passengers under 18:", len(children))

# Multiple conditions: use & and |, and bracket each condition
young_first = df[(df["age"] < 18) & (df["pclass"] == 1)]
print("Under 18 in first class:", len(young_first))
````

`and` / `or` do **not** work here — Pandas needs `&` and `|`, and each condition must sit in its own brackets. This trips up everyone once.

## 6. Count and group

````python
print(df["survived"].value_counts())        # how many of each value
print()
print(df["sex"].value_counts())
````

````python
# The single most useful exploratory line in Pandas:
df.groupby("sex")["survived"].mean()
````

Read it as: split the table by `sex`, take the `survived` column of each group, average it. Since survived is 0/1, the mean *is* the survival rate.

> **Your turn:** survival rate by `pclass`. Then by `sex` and `pclass` together — pass a list: `groupby(["sex", "pclass"])`.

````python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("titanic")

print("Survival rate by pclass:")
print(df.groupby("pclass")["survived"].mean())
print("\nSurvival rate by sex and pclass:")
print(df.groupby(["sex" ,"pclass"])["survived"].mean())# your code
````
